## 1. My lane (or freestyle) and why

I am choosing a freestyle lane focused on ranking content pages for refresh or optimization. My provisional question is which pages should be reviewed first when the team wants to reduce the risk of traffic decline or improve page performance.

I chose this lane because the starter data appears to contain enough page-level signals to support a useful prioritization problem. If the data can help identify pages that are likely to need attention, that could make a real difference in how editorial or SEO time is spent.

```python
import pandas as pd
from pathlib import Path

possible_paths = [
    Path('data/raw/content_refresh_anonymized.csv'),
    Path('data/content_refresh_anonymized.csv'),
    Path('content_refresh_anonymized.csv')
]

csv_path = next((p for p in possible_paths if p.exists()), None)
if csv_path is None:
    raise FileNotFoundError('Could not find the starter CSV. Update the path in this cell to the correct dataset file.')

df = pd.read_csv(csv_path)
df.head()
```

## 2. The question: decision, action, cost of a wrong call

**Search question:** Which content pages should be prioritized for refresh or optimization next?

**Unit of analysis:** One page.

**Output:** A ranked list or priority score for pages.

**Decision this improves:** Which pages should be reviewed or updated first.

**Action someone could take:** An editor, SEO lead, or content operator could use the ranking to decide where to spend time first.

**Cost of a wrong call:** If the ranking is wrong, the team could spend time on pages that do not need help, or miss pages that are actually losing performance.

```python
rows = len(df)
clients = df['client_id'].nunique() if 'client_id' in df.columns else None
cols = df.shape[1]

print(f'Rows: {rows}')
print(f'Clients: {clients}')
print(f'Columns: {cols}')
```

## 3. Quick look at the data (2-3 real numbers)

I will load the starter CSV and show a few real numbers that make this lane worth the next 7 weeks. The goal is not to prove the final solution yet. The goal is to show that the data is large enough and structured enough to make this problem worth exploring.

```python
numbers = {}

numbers['rows'] = len(df)
numbers['clients'] = df['client_id'].nunique() if 'client_id' in df.columns else None
numbers['columns'] = df.shape[1]

for col in ['is_declining', 'needs_ctr_fix', 'needs_engagement_fix']:
    if col in df.columns:
        numbers[f'{col}_rate'] = float(df[col].mean())

for col in ['ctr', 'avg_position', 'engagement_rate', 'health_score', 'trend_pct']:
    if col in df.columns:
        numbers[f'{col}_mean'] = float(df[col].mean())

numbers
```
